# 02 · Megascale 4,096-Qubit Cat+$T$, 2,048-Qubit Stabilizer & 64-Qubit $2^{64}$-Ket `ExactMps`

`VerdantAqmBackend` sets `num_qubits = None` and routes structured circuits to `StabilizerTableau` (`AQ-04`), `BigInt`-indexed `CyclotomicExact` (`AQ-03`), or $\mathbb{Q}(\zeta_8)$ `ExactMps` (`AQ-14`) without dense $2^n$ state-vector allocation.

```python
from qiskit import QuantumCircuit
from qiskit_verdant_aqm import VerdantAqmBackend

n = 4096
cat = QuantumCircuit(n, 2, name="cat_4096q_octant")
cat.h(0)
for q in range(n - 1):
    cat.cx(q, q + 1)
cat.t(0)
cat.t(4095)
cat.measure([0, 4095], [0, 1])

backend = VerdantAqmBackend(shots=1024, seed_simulator=2026)
result = backend.run(cat).result()
```

```python
from qiskit import QuantumCircuit
from qiskit_verdant_aqm import VerdantAqmBackend

n = 2048
stab = QuantumCircuit(n, 2, name="stabilizer_2048q")
stab.h(0)
for q in range(n - 1):
    stab.cx(q, q + 1)
stab.s(0)
stab.sdg(0)
stab.measure([0, 2047], [0, 1])

backend = VerdantAqmBackend(shots=512, seed_simulator=77)
result = backend.run(stab).result()
```

```python
from qiskit import QuantumCircuit
from qiskit_verdant_aqm import VerdantAqmBackend

rev = QuantumCircuit(4, 4, name="toffoli_fredkin_exact")
rev.h(0)
rev.x(1)
rev.ccx(0, 1, 2)
rev.cswap(2, 1, 3)
rev.measure([0, 1, 2, 3], [0, 1, 2, 3])

backend = VerdantAqmBackend(shots=1024, seed_simulator=99)
result = backend.run(rev).result()
```

```python
from qiskit import QuantumCircuit
from qiskit_verdant_aqm import VerdantAqmBackend

n = 64
mps = QuantumCircuit(n, 2, name="cluster_octant_64q_mps")
for q in range(n):
    mps.h(q)
    mps.t(q)
for q in range(n - 1):
    mps.cz(q, q + 1)
mps.measure([0, 63], [0, 1])

backend = VerdantAqmBackend(shots=1024, seed_simulator=314)
result = backend.run(mps).result()
```
